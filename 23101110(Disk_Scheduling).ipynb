FCFS

```python
req_seq = [0, 13, 41, 50, 63, 89, 100]
head = 46
totalseek = 0
for i in req_seq:
  totalseek += abs( i-head )
  head = i
  print("Total head movement:",totalseek)
```

SSTF

```python
requests=[0,13,41,50,63,89,100]
head = 46
tst= 0
ss=[head]
while requests:
 closest = min(requests, key=lambda x: abs(x - head))
 st = abs(head - closest)
 tst += st
 head = closest
 ss.append(head)
 requests.remove(closest)

print("SSTF")
print("Seek Sequence:","->".join(map(str,ss)))
print("Total Seek Time",tst)
```

Scan (Direction-Left)

```python
requests = [0, 13, 41, 50, 63, 89, 100]
head = 46

total = 0

print("Seek Sequence",head,end="->")

right= sorted([x for x in requests if x>=head])
left= sorted([x for x in requests if x<=head],reverse=True)

seek_sequence = []
for x in left:
  total+=abs(head-x)
  head=x
  print(head,end="->")
  seek_sequence.append(x)

for x in right:
  total+=abs(head-x)
  head=x
  print(head,end="->")
  seek_sequence.append(x)

print("\nTotal Seek Time =",total)


```
