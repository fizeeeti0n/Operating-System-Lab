**Priotity Scheduling Non-Preemptive**

```python
processes = [
    ["p1", 3, 4, 2],
    ["p2", 4, 2, 0],
    ["p3", 2, 6, 3],
    ["p4", 0, 3, 6],
    ["p5", 1, 2, 1]
]

n = len(processes)
time = 0
comple = 0
done = [False] *n

CT = [0]*n
TAT = [0]*n
WT = [0]*n

while comple < n:
  available = []
  for i in range(n):
   if processes[i][1]<= time and not done [i]:
    available.append (i)

  if not available:
    time +=1
    continue



  index = min(available, key=lambda i: processes[i][3])
  pid , AT, BT, priority =  processes[index]
  time += BT
  CT[index] = time
  done[index] = True
  comple += 1

  TAT[index] = CT[index] - AT
  WT[index] = TAT[index] - BT

print("Process\tAT\tBT\tPriority\tCT\tTAT\tWT")
total_wt = 0
total_tat = 0
for i in range(n):

    print(processes[i][0], "\t", processes[i][1], "\t",

          processes[i][2], "\t", processes[i][3], "\t\t",

          CT[i], "\t", TAT[i], "\t", WT[i])

    total_wt += WT[i]

    total_tat += TAT[i]


print(f"Average TAT: {sum(TAT)/n}")
print(f"Average WT: {sum(WT)/n}")
```

**Priotity Scheduling Preemptive**

```python
processes = [
    ["p1", 3, 4, 2],
    ["p2", 4, 2, 0],
    ["p3", 2, 6, 3],
    ["p4", 0, 3, 6],
    ["p5", 1, 2, 1]
]


n = len(processes)
RemainingTime = [p[2] for p in processes]
CompletionTime = [0] * n
TurnAroundTime = [0] * n
WaitingTime = [0] * n
Time = 0
Completed = 0

while Completed < n:
    Available = []
    for i in range(n):
        if processes[i][1] <= Time and RemainingTime[i] > 0:
            Available.append(i)

    if not Available:
        Time += 1
        continue

    index = min(Available, key=lambda i: (processes[i][3], processes[i][1]))

    RemainingTime[index] -= 1
    Time += 1

    if RemainingTime[index] == 0:
        CompletionTime[index] = Time
        TurnAroundTime[index] = CompletionTime[index] - processes[index][1]
        WaitingTime[index] = TurnAroundTime[index] - processes[index][2]
        Completed += 1

print("Process\tAT\tBT\tPriority\tCT\tTAT\tWT")
total_wt = 0
total_tat = 0
for i in range(n):
    print(processes[i][0], "\t", processes[i][1], "\t",
          processes[i][2], "\t", processes[i][3], "\t\t",
          CompletionTime[i], "\t", TurnAroundTime[i], "\t", WaitingTime[i])
    total_wt += WaitingTime[i]
    total_tat += TurnAroundTime[i]

print(f"Average TAT: {total_tat/n}")
print(f"Average WT: {total_wt/n}")
```
